# Step 4: Fail Detection
Find students who failed.
- Condition: marks < 35 in any subject
# Concepts used:
- Boolean masking
- np.where
# Tasks:
- count failed students
- list their indexes

```python
fail_students=np.where(marks<35)
fail_students
# How to read this output : in this output there are two array : first is row array and and second is column array.It has returned the output for
# example  0th row and 1 st column is where the marks is lesss than 35.
```

```python
## To make this more readable we will use , this boolean masking '
fail_stud=marks<35
fail_stud
```

```python
# finding out the students which have failed (any subject)
fail_stud=np.any(marks<35,axis=1)
fail_stud
```

```python
# finding out the index on which the students  failed 
failed_index=np.where(fail_stud)[0]
failed_index
```

# Summary (for notebook):

np.where(condition) returns a tuple of arrays containing the indexes where the condition is True.

For a 1-D array, it returns something like:

(array([1, 2, 4]),)

The result is a tuple with one element (the index array).

If we use:

failed_indexes = np.where(failed_students)

then:

len(failed_indexes) = 1

because len() counts elements in the tuple, not the indexes.

To get the actual index array we use [0]:

failed_indexes = np.where(failed_students)[0]

Now:

failed_indexes → [1 2 4]
len(failed_indexes) → 3

✅ Conclusion:
[0] extracts the index array from the tuple returned by np.where(). This allows len() to correctly count the number of failed students.
# In plain language [0] is called accessing the elements at index zero. 

```python
# Counting how may students failded 
failed_count=len(failed_index)
failed_count
```

# Step 5: Grade Assignment

```python
# Publishing the grade of the using the np.select
conditions=[(marks>=90),(marks<=89) & (marks>=75),(marks<=74) &( marks>=60),(marks<=59 )&(marks>=40),(marks<40)]
grade=['A','B','C','D','F']
result=np.select(conditions,grade,default="F")
result
```

```python
# publishing the grade using the boolean masking 
grades = np.empty(marks.shape, dtype=str)
grades[marks >= 90] = 'A'
grades[(marks >= 75) & (marks < 90)] = 'B'
grades[(marks >= 60) & (marks < 75)] = 'C'
grades[(marks >= 40) & (marks < 60)] = 'D'
grades[marks < 40] = 'F'

print(grades)
```

 # Step 7: Score Normalization (Advanced)
- Apply Min-Max normalization:
          - X_norm = (X - min)/(max-min)
- Concepts used:
  - broadcasting
  - vectorization
    
Score Normalization means scaling values so they fall into a common range, usually 0 to 1.
It does not change the relative order of scores, it only rescales them.
This is widely used in data science, machine learning, and statistics.
- Why It Is Important (Real Use)
-     - Normalization is used when:
             - 1️⃣ Machine Learning models (very important)
             - 2️⃣ Feature scaling
             - 3️⃣ Comparing variables with different ranges

```python
marks_norm=(marks-marks.min())/(marks.max()-marks.min())
marks_norm
```
