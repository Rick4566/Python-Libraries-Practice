# Main Data Structures in Pandas

Pandas has two primary data structures:

## 1. Series

### Definition
A Series is a one-dimensional labeled array capable of holding data of any type.

### Features
- One-dimensional
- Has index labels
- Stores single column data
- Mutable in size

### Example

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40])
print(s)

# 2. DataFrame
Definition

A DataFrame is a two-dimensional labeled data structure with rows and columns.

# Features
Two-dimensional
Tabular format
Rows and columns
Can store multiple data types
Most commonly used Pandas structure

````python
import numpy as np
import pandas as pd
df=pd.read_csv('matches - matches.csv')
````

````python
# You can apply fancing indexing also to fetch discontinious row : ex fetch 1, 5,6 row
df.iloc[[1,5,6]]
````

````python
#BY iloc you can also fetch coulumns : to get the all row use colon , then apply fancy indexing list 
# and pass the index value of the column .It is very powerful
df.iloc[:,[4,5,10]]
````

````python
#Masking : “Give me rows from data where the condition is True.”
# This technique is called Boolean indexing or masking in pandas.
mask=df['city']=='Hyderabad'
df[mask]
````

````python
#Data analyst person create the function like this to get the value and perform analysis
def get_city_match_count(city):
    mask=df['city']==city
    return df[mask].shape[0]
````

````python
get_city_match_count("Hyderabad")
get_city_match_count("Rajkot")
````

````python
mask1=df['city']=='Hyderabad'
mask2=df['date']>'2010-01-01'
df[mask1 & mask2] # Here & is used instead of "and" because mask1 and mask2 are boolean series 
# and in boolean series we apply bitwise and  & not normal and .
````

````python
# if we want to find count
df[mask1 & mask2].shape[0]
````

 # Level 2 — Filtering and conditions

````python
#Display all matches played in the 2017 season.
df[df["season"]==2017]
````

````python
#Find all matches played in Hyderabad.
df[df["city"]=="Hyderabad"]
````

````python
#Display matches where team1 was Mumbai Indians.
df[df["team1"]=="Mumbai Indians"]
````

````python
#Find matches won by Chennai Super Kings.
df[df["winner"]=="Chennai Super Kings"]
````

````python
#Find matches where the winning margin in runs was greater than 50.
df[df["win_by_runs"]>50]
````

````python
#Find matches decided by 10 wickets.
df[df["win_by_wickets"]==10]
````

````python
#Display matches where the toss winner also won the match.
df[df["toss_winner"]==df["winner"]]
````

````python
#Find matches where the toss decision was bat.
df[df["toss_decision"]=="bat"]
````

````python
#Find matches played in either Mumbai or Kolkata.
df[(df["city"]=="Mumbai" )|( df["city"]=="kolkata")]
````

````python
#Find matches where the winner was Mumbai Indians and the match was played in the 2019 season.
df[(df["winner"]=="Mumbai Indians") & (df["season"]==2017)]
````

````python
#Find matches where player_of_match was missing.
df[df["player_of_match"].isnull()]
````

````python
#Find matches where the toss winner and match winner were different.
df[df["toss_winner"] != df["winner"]]
````
